```python
import json


```

```python
# Load the original JSON
with open('filtered_records_on_zeolite.json', 'r') as file:
    filtered_records = json.load(file)

# Remove 'paragraphs' field from each record
cleaned_records = []
for record in filtered_records:
    # Make a shallow copy so we don't mutate the original data
    new_record = {key: value for key, value in record.items() if key != "paragraphs"}
    cleaned_records.append(new_record)

# Save to a new JSON file
with open('abstract_only_records.json', 'w') as outfile:
    json.dump(cleaned_records, outfile, indent=2)

print(f"Saved {len(cleaned_records)} records with 'paragraphs' removed.")
```
