```python
#input file is the overall zeolite paper database with all publishers and it is filtered to random 100 entries

from pathlib import Path
import json

SOURCE_PATH = Path("filtered_records_on_zeolite.json")
OUTPUT_PATH = Path("filtered_records_subset_100.json")
```

```python
with SOURCE_PATH.open("r", encoding="utf-8") as f:
    records = json.load(f)

if not isinstance(records, list):
    raise TypeError("Expected the source file to contain a list of records.")

print(f"Loaded {len(records)} records from {SOURCE_PATH}")
```

```python
subset_size = min(100, len(records))
subset = records[:subset_size]

with OUTPUT_PATH.open("w", encoding="utf-8") as f:
    json.dump(subset, f, indent=2)

print(f"Saved {subset_size} records to {OUTPUT_PATH}")
```
