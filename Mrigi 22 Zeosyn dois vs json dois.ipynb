```python
import json
import pandas as pd

import pandas as pd

# Load the Excel file to see sheet names and structure
excel_path = "ZEOSYN.xlsx"
xlsx = pd.ExcelFile(excel_path)
xlsx.sheet_names
```

```python
# Load the contents of the sheet
df = pd.read_excel(excel_path, sheet_name="Sheet1")

# Display the first few rows
df.head()
```

```python
# Count total number of unique DOIs
total_dois = len(df['doi'].unique())
print(f"Total unique DOIs: {total_dois}")

# Find duplicate DOIs
duplicate_dois = df[df['doi'].duplicated(keep=False)].sort_values('doi')
if len(duplicate_dois) > 0:
    print(f"\nNumber of rows with duplicate DOIs: {len(duplicate_dois)}")
    print("\nFirst few duplicate DOIs:")
    print(duplicate_dois['doi'].value_counts().head())
else:
    print("\nNo duplicate DOIs found")
```

```python
# Load and analyze the JSON file
with open('filtered_records_on_additionalkeywords.json', 'r') as f:
    records = json.load(f)
    
print(f"Number of entries in JSON file: {len(records)}")
print("\nFirst entry:")
print(json.dumps(records[0], indent=2))
```

```python
# Extract DOIs from records list
json_dois = set(record['doi'] for record in records)

# Compare with Excel DOIs
excel_dois = set(df['doi'])

# Print statistics
print(f"Number of DOIs in JSON: {len(json_dois)}")
print(f"Number of DOIs in Excel: {len(excel_dois)}")

# Find duplicates
duplicates = json_dois.intersection(excel_dois)
print(f"\nNumber of duplicate DOIs between JSON and Excel: {len(duplicates)}")

if len(duplicates) > 0:
    print("\nFirst 5 duplicate DOIs:")
    for doi in list(duplicates)[:5]:
        print(doi)
```

```python
# Load the new JSON file
with open('filtered_records_on_zeolite.json', 'r') as f:
    zeolite_records = json.load(f)

# Extract DOIs from zeolite records
zeolite_dois = set(record['doi'] for record in zeolite_records)

# Compare with existing JSON dois
print(f"Number of DOIs in zeolite JSON: {len(zeolite_dois)}")
print(f"Number of DOIs in keywords JSON: {len(json_dois)}")

# Find overlapping DOIs
overlapping_dois = zeolite_dois.intersection(json_dois)
print(f"\nNumber of overlapping DOIs between the two JSON files: {len(overlapping_dois)}")

if len(overlapping_dois) > 0:
    print("\nFirst 5 overlapping DOIs:")
    for doi in list(overlapping_dois)[:5]:
        print(doi)

# Compare with Excel DOIs as well
excel_overlapping = zeolite_dois.intersection(excel_dois)
print(f"\nNumber of DOIs from zeolite JSON that overlap with Excel: {len(excel_overlapping)}")
```

```python
# DOIs from Excel only (not in either JSON)
excel_only = excel_dois - (json_dois | zeolite_dois)

# DOIs in first JSON only (not in Excel or zeolite JSON)
json_only = json_dois - (excel_dois | zeolite_dois)

# DOIs in zeolite JSON only (not in Excel or first JSON)
zeolite_only = zeolite_dois - (excel_dois | json_dois)

# DOIs in both JSONs but not Excel
both_jsons_not_excel = json_dois.intersection(zeolite_dois) - excel_dois

# DOIs in all three sources
in_all = excel_dois.intersection(json_dois).intersection(zeolite_dois)

print(f"DOIs in Excel only: {len(excel_only)}")
print(f"DOIs in first JSON only: {len(json_only)}")
print(f"DOIs in zeolite JSON only: {len(zeolite_only)}")
print(f"DOIs in both JSONs but not Excel: {len(both_jsons_not_excel)}")
print(f"DOIs in all three sources: {len(in_all)}")

if len(excel_only) > 0:
    print("\nFirst 5 DOIs in Excel only:")
    for doi in list(excel_only)[:5]:
        print(doi)
```

```python
# Create a new dataframe with only the rows that have DOIs present in zeolite_dois
filtered_df = df[df['doi'].isin(zeolite_dois)]

# Save the filtered dataframe to a new Excel file
filtered_df.to_excel('filtered_ZEOSYN.xlsx', index=False)

# Print some statistics
print(f"Original number of rows: {len(df)}")
print(f"Number of rows after filtering: {len(filtered_df)}")
print(f"Number of rows removed: {len(df) - len(filtered_df)}")
```
