```python
import pandas as pd

# Load the Excel file to see sheet names and structure
excel_path = "ZEOSYN.xlsx"
xlsx = pd.ExcelFile(excel_path)
xlsx.sheet_names
```

```python
# Load the contents of the sheet
df = pd.read_excel(excel_path, sheet_name="Sheet1")

# Display the first few rows
df.head()
```

```python
import json

# Extract unique non-null values from the code columns
code_columns = ['Code1', 'Code2', 'Code3']
product_codes = pd.unique(df[code_columns].values.ravel('K'))
product_codes = [code for code in product_codes if pd.notna(code)]

# Sort and save as JSON
product_codes = sorted(set(product_codes))
json_path = "zeolite_product_codes.json"
with open(json_path, 'w') as f:
    json.dump(product_codes, f, indent=2)

product_codes[:10], f"Total unique codes: {len(product_codes)}", json_path
```
